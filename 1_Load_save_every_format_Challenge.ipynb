## Json dosyalarıyla çalışma

Aşağıdaki hücreyi çalıştırarak `d` sözlüğünü başlatın.

```python
d = {"firstname": "leonardo",
     "lastname": "dicaprio",
     "job": "actor"}
```

`d` klasöründen `actors.json` adlı bir dosya oluşturun. Sözdizimi konusunda kafanız karıştıysa ders notlarına bakabilirsiniz.

```python
import json

with open ("/content/drive/MyDrive/Colab Notebooks/actors.json", "w") as f:
  json.dump(d, f, ensure_ascii=False, indent=2)
```

Yeni `actors.json` dosyasını açın ve yeni bir anahtar `nationality` ekleyin (Leonardo Dicaprio Amerikalıdır). \
Sonucu aynı dosya adına kaydedin. Dosyanızı kontrol edin.

```python
import json

with open("/content/drive/MyDrive/Colab Notebooks/actors.json", "r", encoding="utf-8") as f:
    data = json.load(f)

data["nationality"] = "American"

with open("/content/drive/MyDrive/Colab Notebooks/actors.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(data)


```

Bu veri çerçevesinde olağandışı bir şey fark ettiniz mi? Excel dosyasındaki sayfayı tekrar inceleyin ve bu garip formatın nedenini anlamaya çalışın.

İkinci sayfadaki (Restaurant2) veriler düzgün formatta değildir.
İlk birkaç satır tamamen boş (NaN) olarak gelmiştir ve gerçek sütun başlıkları (Date, Turnover, Cost, Margin) üçüncü satırda yer almaktadır.
Bu nedenle Pandas sütun isimlerini Unnamed: 0, Unnamed: 1 gibi otomatik adlandırmıştır.
Veriyi doğru şekilde yüklemek için skiprows veya header parametresi kullanılarak başlık satırı doğru şekilde tanımlanmalıdır.


Verileri doğru şekilde yüklemek için önceki kod satırınızı değiştirin. \
Şüpheleriniz varsa [belgelere](https://pandas.pydata.org/docs/reference/api/pandas.read_excel.html) göz atın.

```python
df_sales_restaurant_2 = pd.read_excel(
    "/content/drive/MyDrive/Colab Notebooks/restaurants_accountability.xlsx",
    sheet_name="Restaurant2",
    skiprows=2
)

df_sales_restaurant_2.head()
```

```python
# Boş ilk iki sütunu at
df_sales_restaurant_2 = df_sales_restaurant_2.iloc[:, 2:]

# Kolon isimlerini düzelt
df_sales_restaurant_2.columns = ["Date", "Turnover", "Cost", "Margin"]

df_sales_restaurant_2.head()
```

Her veri çerçevesine yeni bir `restaurant_id` sütunu ekleyin (restaurant_1 için id = 1 ve restaurant_2 için id = 2) ve iki veri çerçevesini `df_sales` adlı tek bir veri çerçevesi halinde birleştirin.


İki veri çerçevesini birleştireceğiniz `eksen`e dikkat edin.

```python
df_sales_restaurant_1["restaurant_id"] = 1
df_sales_restaurant_2["restaurant_id"] = 2


```

```python
df_sales = pd.concat(
    [df_sales_restaurant_1, df_sales_restaurant_2],
    axis=0,
    ignore_index=True
)

df_sales.head()
```

Birleştirilmiş veri çerçevesini, `All` adlı bir sayfada yeni bir `restaurants_accountability_full.xlsx` dosyasına aktarın.

```python

output_path = "/content/drive/MyDrive/Colab Notebooks/restaurants_accountability_full.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    df_sales.to_excel(writer, sheet_name="ALL", index=False)

print("Kaydedildi:", output_path)
```
