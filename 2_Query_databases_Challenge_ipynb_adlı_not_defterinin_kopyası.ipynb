```python
from google.cloud import bigquery
import pandas as pd

# Bigquery client
client = bigquery.Client(project = "deniz-challenge1")

#SQL sorgusu
query = """
SELECT *
FROM `deniz-challenge1.course29.gwz_orders_samples`

"""

#DataFrame'e yükle
df = client.query(query).to_dataframe()

#Kontrol
df.head()

```
